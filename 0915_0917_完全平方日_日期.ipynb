```python
def is_leap_year(year):
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False

def is_date_valid(date):
    day = date % 100
    remain = date // 100
    month = remain % 100
    year = remain // 100
    if year >= 1:
        if 12 >= month >= 1:
            day_per_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
            max_days = day_per_month[month-1]
            # if month == 2 and is_leap_year(year) == True
            if month == 2 and is_leap_year(year):
                max_days = max_days + 1
            if max_days >= day >= 1:
                return True
            else:
                return False
        else:
            return False
    else:
        return False

def catch_perfect_square_day(start=4500, maxcount=10):
    count = 0
    result = []
    while True:
        d = start * start
        if is_date_valid(d):
            result.append(d)
            count = count + 1
        if count == maxcount:
            break
        start = start + 1
    return result

print(catch_perfect_square_day())
print(catch_perfect_square_day(5555,5))
```

```python
# python的日期時間模組有兩個
# 1. time: 時間
# 2. datetime: 日期+時間
# str<->datetime 型態轉換
# strptime: str -> datetime
# strftime: datetime -> str
import datetime as dt
s = "2026-09-15"
d = dt.datetime.strptime(s, "%Y-%m-%d")
print(d)

# 專屬功能
s2 = d.strftime("%m/%d(%a)")
print(s2)
```

```python
import datetime as dt

# 建立日期：2026/9/15
d = dt.datetime(2026, 9, 15)
print(d)

# 建立一段時間：1週 + 1天 = 8天
delta = dt.timedelta(days=1, weeks=1)

# 往前推 2 個 delta，也就是 16 天
d = d - 2 * delta

print(d)
# 2026-08-30 00:00:00
```

```python
# 想法1. 迴圈
import datetime as dt

def catch_perfect_root_day(y=2024, m=10, d=1, maxcount=10):
    d = dt.datetime(y, m, d)
    delta = dt.timedelta(days=1)
    count = 0
    result = []
    while True:
        # 一天往後數
        # 開根號(**0.5)->確認是否是整數
        d_str = d.strftime("%Y%m%d")
        d_int = int(d_str)
        d_root = d_int ** 0.5
        if d_root.is_integer():
            # 如果是整數, 就是個完全平方日
            result.append(d_int)
            count = count + 1
            if count == maxcount:
                break
        d = d + delta
    return result

print(catch_perfect_root_day())
print(catch_perfect_root_day(3100,5,2,3))
```
