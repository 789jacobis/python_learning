```python
#ABCDABCD
#BCDABCDA
#CDABCDAB
#DABCDABC

list1 = ["A", "B", "C", "D", "E"]

for i in range(len(list1)):
    for j in range(len(list1)*2):
        print(list1[(i+j)%len(list1)], end="")
    print()

# i=0, j=0 → list1[(0+0)%5] = list1[0] = A → ABCDEABCDE
# i=1, j=0 → list1[(1+0)%5] = list1[1] = B → BCDEABCDEA
# i=2, j=0 → list1[(2+0)%5] = list1[2] = C → CDEABCDEAB
# i=3, j=0 → list1[(3+0)%5] = list1[3] = D → DEABCDEABC
# i=4, j=0 → list1[(4+0)%5] = list1[4] = E → EABCDEABCD

# 每列 j 都會從 0 跑到 9
# (i+j)%5 讓索引超過 4 後重新回到 0
```
