```python
for i in range(4):
    line = ""
    for j in range(5):
        if (j + i) % 3 == 0:
            line = line + "o"
        elif (j + i) % 3 == 1:
            line = line + "x"
        elif (j + i) % 3 == 2:
            line = line + "!"
    print(line)

print()

symbols = ["o", "x", "!"]
for i in range(4):
    line = ""
    for j in range(5):
        pos = (j + i) % len(symbols)
        line = line + symbols[pos]
    print(line)
```

```python
#ABCDABCD
#BCDABCDA
#CDABCDAB
#DABCDABC

list1 = ["A", "B", "C", "D", "E"]

for i in range(len(list1)):
    for j in range(len(list1)*2):
        print(list1[(i+j)%len(list1)], end="")
    print()

# i=0, j=0 → list1[(0+0)%5] = list1[0] = A → ABCDEABCDE
# i=1, j=0 → list1[(1+0)%5] = list1[1] = B → BCDEABCDEA
# i=2, j=0 → list1[(2+0)%5] = list1[2] = C → CDEABCDEAB
# i=3, j=0 → list1[(3+0)%5] = list1[3] = D → DEABCDEABC
# i=4, j=0 → list1[(4+0)%5] = list1[4] = E → EABCDEABCD

# 每列 j 都會從 0 跑到 9
# (i+j)%5 讓索引超過 4 後重新回到 0
```

```python
# 初始化一個list
# 方法1.
a = [0] * 3
print(a)
# 方法2. 初始化, 順手填值(!)
a = []
for i in range(5):
    a.append(i)
print(a)
# 方法3. 把方法2精簡
a = [(i + 1) for i in range(5)]
print(a)
```

```python
# 二維list的初始化
# 錯的, 因為你b只創造了一次, 你就算裡面有三個, 還是指同一個
b = [0] * 5
a = [b] * 3
print(a)
a[0][0] = 99
print(a)
# 方法2
a = []
for i in range(3):
    b = [0] * 5
    a.append(b)
a[0][0] = 88
print(a)
# 方法3
a = [[0] * 5 for i in range(3)]
a[0][0] = 77
print(a)
```
