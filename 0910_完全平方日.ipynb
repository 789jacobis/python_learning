```python
basic_number = 4500
i = 0
count = 0

def modulo(number):
    a = (number // 10000000) % 10 #年/0
    b = (number // 1000000) % 10  #年/1
    c = (number // 100000) % 10   #年/2
    d = (number // 10000) % 10    #年/3
    e = (number // 1000) % 10     #月/4
    f = (number // 100) % 10      #月/5
    g = (number // 10) % 10       #日/6
    h = number % 10               #日/7
    return a, b, c, d, e, f, g, h

def check_leap_year(year):
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False

while count < 10:
    test_date = (basic_number + i) ** 2
    # 如果是閏年
    if check_leap_year(modulo(test_date)[0] * 1000 + modulo(test_date)[1] * 100 + modulo(test_date)[2] * 10 + modulo(test_date)[3]):
        # 先判定月份必須介於1~12之間再去思考後續步驟
        if 1 <= ((modulo(test_date)[4] * 10) + modulo(test_date)[5]) <= 12:
        # 如果是大月
            if ((modulo(test_date)[4] * 10) + modulo(test_date)[5]) in [1, 3, 5, 7, 8, 10, 12]:
                # 日期範圍為: 1~31
                if 1 <= modulo(test_date)[6] * 10 + modulo(test_date)[7] <= 31:
                    print(f"{test_date}這一天是完全平方日")
                    count += 1
            # 如果是小月
            else:
                # 如果是2月
                if modulo(test_date)[4] * 10 + modulo(test_date)[5] == 2:
                    if 1 <= modulo(test_date)[6] * 10 + modulo(test_date)[7] <= 29:
                        print(f"{test_date}這一天是完全平方日")
                        count += 1
                # 如果不是2月
                else:
                    if 1 <= modulo(test_date)[6] * 10 + modulo(test_date)[7] <= 30:
                        print(f"{test_date}這一天是完全平方日")
                        count += 1
    # 如果不是閏年
    else:
        # 先判定月份必須介於1~12之間再去思考後續步驟
        if 1 <= ((modulo(test_date)[4] * 10) + modulo(test_date)[5]) <= 12:
            # 如果是大月
            if ((modulo(test_date)[4] * 10) + modulo(test_date)[5]) in [1, 3, 5, 7, 8, 10, 12]:
                # 日期範圍為: 1~31
                if 1 <= modulo(test_date)[6] * 10 + modulo(test_date)[7] <= 31:
                    print(f"{test_date}這一天是完全平方日")
                    count += 1
            # 如果是小月
            else:
                # 如果是2月
                if modulo(test_date)[4] * 10 + modulo(test_date)[5] == 2:
                    if 1 <= modulo(test_date)[6] * 10 + modulo(test_date)[7] <= 28:
                        print(f"{test_date}這一天是完全平方日")
                        count += 1
                # 如果不是2月
                else:
                    if 1 <= modulo(test_date)[6] * 10 + modulo(test_date)[7] <= 30:
                        print(f"{test_date}這一天是完全平方日")
                        count += 1

    i += 1


```
